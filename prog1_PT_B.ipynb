```python
def leaky_bucket():
    drop = 0
    count = 0

    cap = int(input("Enter the bucket size: "))
    process = int(input("Enter the processing rate: "))
    nsec = int(input("Enter the number of seconds: "))

    inp = []

    for i in range(nsec):
        packet = int(input(f"Enter packet size at {i + 1} sec: "))
        inp.append(packet)

    print("\nSecond | Received | Sent | Packet Left | Dropped")
    print("------------------------------------------------")

    for i in range(nsec):
        count += inp[i]

        if count > cap:
            drop = count - cap
            count = cap

        mini = min(count, process)
        count -= mini

        print(f"{i+1}\t {inp[i]}\t   {mini}\t    {count}\t       {drop}")
        drop = 0

    while count > 0:
        mini = min(count, process)
        count -= mini

        print(f"{i+2}\t 0\t   {mini}\t    {count}\t       {drop}")
        i += 1


leaky_bucket()
```
